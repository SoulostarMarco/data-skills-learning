**6. 统计计算 —— 收益率分析**

给定某股票 10 天的日收益率（%）：
```python
returns = np.array([1.2, -0.5, 0.8, -1.0, 2.5, 0.3, -0.2, 1.5, -0.8, 0.6])
```
- 计算平均日收益率
- 计算收益率的标准差（波动率）
- 计算累计收益率（提示：`cumsum`）
- 找出收益率最高和最低的日期索引

````python
# 你的代码
returns = np.array([1.2, -0.5, 0.8, -1.0, 2.5, 0.3, -0.2, 1.5, -0.8, 0.6])

print(np.mean(returns))
print(np.std(returns))
print(np.cumsum(returns))
print(np.argmax(returns))
print(np.argmin(returns))
print(np.mean(returns[returns > 0]))
print(returns[returns > 0].mean())
````

参考答案

````python
# 方法1：布尔数组的 mean 就是 True 的比例
print((returns > 0).mean())  # 0.6（6天正收益 / 10天）

# 方法2：显式计数
print(np.sum(returns > 0) / len(returns))  # 0.6
````

**7. 条件替换与复合筛选**

给定数组 `data = np.array([15, 8, 22, 3, 18, 9, 30, 12, 5, 25])`：
- 创建一个新数组，规则：>20 的变为 '高'，<10 的变为 '低'，其余变为 '中'
  （提示：先创建字符串数组 `np.full_like`，或用 `np.where` 嵌套）
- 统计 '高'、'中'、'低' 各有多少个

**8. 矩阵运算 —— 投资组合权重**

给定 3 只股票 × 5 天的收益率矩阵（行=股票，列=天数）：
```python
returns_matrix = np.array([
    [0.01,  0.02, -0.01,  0.015,  0.005],   # 股票A
    [0.005, 0.01,  0.02, -0.005,  0.01 ],   # 股票B
    [0.02, -0.01,  0.01,  0.02,  -0.005]    # 股票C
])
weights = np.array([0.5, 0.3, 0.2])  # 投资权重
```
- 计算每只股票 5 天的平均收益率（axis=?）
- 用**广播**计算：每天的投资组合收益率 = 当天3只股票的收益率 × 权重后求和
  （提示：`returns_matrix * weights.reshape(3,1)` 或直接用广播）
- 计算投资组合 5 天的平均收益率和波动率（标准差）
- 找出投资组合收益最高的是第几天

````python
# 你的代码
returns_matrix = np.array([
    [0.01,  0.02, -0.01,  0.015,  0.005],   # 股票A
    [0.005, 0.01,  0.02, -0.005,  0.01 ],   # 股票B
    [0.02, -0.01,  0.01,  0.02,  -0.005]    # 股票C
])
weights = np.array([0.5, 0.3, 0.2])  # 投资权重

print(returns_matrix.mean(axis=1))
portfolio = returns_matrix * weights.reshape(3, 1)
print(portfolio.sum(axis=0, keepdims=True))
print(np.mean(portfolio))
print(np.std(portfolio))
print(portfolio.sum(axis=0, keepdims=True).argmax())
````

参考答案

问题: portfolio 是加权后的 3×5 矩阵，但「投资组合收益率」应该是每天一个值（3只股票当天收益按权重加权求和），是一个长度为 5 的一维数组。

你算 np.mean(portfolio) 得到的是 15 个元素的平均，不是「投资组合 5 天的平均收益率」。

````python
# 每天的投资组合收益率（一维数组，长度=5）
daily_portfolio = (returns_matrix * weights.reshape(3, 1)).sum(axis=0)
# 或: daily_portfolio = (returns_matrix.T @ weights)  # 矩阵乘法写法

print(daily_portfolio)  # [0.0105 0.011  0.003  0.01   0.0045]

# 投资组合的平均收益率和波动率
print(daily_portfolio.mean())   # 0.0078
print(daily_portfolio.std())    # 0.00344

# 收益最高的是第几天
print(daily_portfolio.argmax())  # 1（第2天，索引从0开始）
````

````python
# 边界测试清单
test_cases = [
    ("空数组", np.array([]), "sum返回0，mean会报Warning/返回nan，需特殊处理"),
    ("全0数组", np.zeros((3,3)), "均值=0，标准差=0，Z-score会除零报错"),
    ("含负数", np.array([-10, 0, 10]), "向量化、布尔索引、统计函数都正常"),
    ("单元素", np.array([[5]]), "std返回0（或nan，取决于ddof）"),
]

for name, arr, expected in test_cases:
    print(f"\n【{name}】")
    print(f"  数据: {arr}")
    try:
        print(f"  sum: {arr.sum()}")
        print(f"  mean: {arr.mean()}")
        print(f"  std: {arr.std()}")
    except Exception as e:
        print(f"  异常: {e}")
````

关键边界教训:
