# 列表推导式深入 + lambda + 高阶函数（练习）

```python
# 1. 列表推导式 —— 从简单到复杂
# 基础形态：[表达式 for 变量 in 序列]
squares = [n**2 for n in range(5)]
print(squares)

# 带过滤：[表达式 for 变量 in 序列 if 条件]
evens = [n for n in range(10) if n % 2 == 0]
print(evens)

# 关键区分：if 在 for 前 vs if 在 for 后
# if 在 for 后 → 过滤（决定保不保留这个元素）
# if 在 for 前 → 三元表达式（决定输出什么）

# 过滤（只保留整数）
result1 = [n for n in range(5) if n % 2 == 0]
print(result1)

# 三元表达式（每个元素都保留，但偶数原样，奇数变'odd'）
result2 = [n if n % 2 == 0 else 'odd' for n in range(5)]
print(result2)

# 同时使用（先过滤再变化）
result3 = [n*10 if n > 2 else n for n in range(5) if n != 0]
# 解读：对range(5) 排除0之后，大于2的乘10，否则原样
print(result3)
```

```python
# 2. 多层嵌套（双重for循环）
# 笛卡尔积：两个list所有组合
colors = ['red', 'blue']
sizes = ['S', 'M', 'L']
products = [f"{c}-{s}" for c in colors for s in sizes]
# 记忆法：for 的顺序和普通双重循环写法完全一致
print(products)

# 扁平化嵌套list
nested = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
flat = [x for sub in nested for x in sub]
print(flat)
```

```python
# 3. 字典推导 + 集合推导
# 字典推导
name_length = {name: len(name) for name in ['Alice', 'Bob', 'Charlie']}
print(name_length)

# 字典推导 + 条件
short_names = {name: len(name) for name in ['Alice', 'Bob', 'Charlie'] if len(name) < 5}
print(short_names)

# 集合推导
unique_length = {len(name) for name in ['Alice', 'Bob', 'Charlie', 'Tom']}
print(unique_length)
```

```python
# 4. lambda —— 匿名函数
# lambda的本质：一行的函数，不需要def
# 等价的两种写法：
def square_v1(x):
    return x**2

square_v2 = lambda x: x**2

print(square_v1(5), square_v2(5))

# lambda的真正用途：作为“参数”传给其他函数
# 不要这样用：
# bad = lambda x: x + 1 
# 这种就该直接用def，反而清晰
```

```python
# 5. sorted的key参数（高频考点）
# 按字符串长度排序
words = ['banana', 'apple', 'kiwi', 'strawberry']
print(sorted(words, key=lambda w: len(w)))

# 按tuple的第二个元素排序
students = [('Alice', 85), ('Bob', 72), ('Charlie', 90)]
print(sorted(students, key=lambda x: x[1], reverse=True))

# 多关键字排序（数据岗高频）：先按A升序，A相同时按B降序
data = [('Alice', 'CS', 85), ('Bob', 'Math', 72), ('Charlie', 'CS', 90)]
# 先按专业升序，再按分数降序
result = sorted(data, key=lambda x: (x[1], -x[2]))
# 技巧：加个负号实现降序
print(result)
```

```python
# 6. map和filter
# map(函数,序列)：对序列每个元素应用函数
# 注意：map返回的是迭代器，要用list()转一下
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, nums))
print(squared)

# filter(函数,序列)：保留函数返回True的元素
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)

# 但是Python社区更推崇列表推导式
# 那为什么还要学 map/filter?
# 因为 Pandas 的 .apply() 本质就是 map 的思想,
# 而面试官有时会问你 map/filter 是什么 → 必须知道
```

```python
# 7. lambda的常见陷阱
# 陷阱1：lambda在循环中捕获变量（经典坑）
funcs = [lambda x: x + i for i in range(3)]
# 你以为是[0, 1, 2], 实际上...
print([f(10) for f in funcs])
# [12, 12, 12] → 全部用了i的最终值

# 修复：用默认参数“冻结”当前的i
funcs_fixed = [lambda x, i=i: x + i for i in range(3)]
print([f(10) for f in funcs_fixed])

# 陷阱2：lambda不能写多行语句
# bad = lambda x: print(x); return x # 语法错误！
# 如果逻辑超过一行，老老实实用def
```
