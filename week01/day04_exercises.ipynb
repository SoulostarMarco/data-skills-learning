# SQL入门 —— SELECT/WHERE/ORDER BY/LIMIT（习题）

```python
import duckdb

%load_ext sql
%sql duckdb://
```

Easy

1. 查询 sales.csv 的前 10 行,只显示 order_id, product, total 三列。

```python
duckdb.sql("""
    SELECT order_id, product, total
    FROM '../data/sales.csv'
    LIMIT 10
""")
```

2. 查询所有 country = 'UK' 的订单的 order_id, customer_id, total。

```python
duckdb.sql("""
    SELECT order_id, customer_id, total
    FROM '../data/sales.csv'
    WHERE country = 'UK'
""")
```

3. 查询 total > 5000 的所有订单,显示 order_id, product, quantity, price, total,按 total 降序排列。

```python
duckdb.sql("""
    SELECT order_id, product, quantity, price, total
    FROM '../data/sales.csv'
    WHERE total > 5000
    ORDER BY total DESC
""")
```

4. 查询有哪几种不同的 product(用 DISTINCT)。

```python
duckdb.sql("""
    SELECT DISTINCT product
    FROM '../data/sales.csv'
""")
```

Medium

5. 查询 country IN ('UK', 'US') 且 category = 'Computer' 的所有订单,显示 order_id, country, product, total,按 country 升序、total 降序排列。
💡 思考:这道题对应昨天 Day 3 第 9 题(订单过滤排序),是不是一模一样的逻辑?

```python
duckdb.sql("""
    SELECT order_id, country, product, total
    FROM '../data/sales.csv'
    WHERE country IN ('UK', 'US') AND category = 'Computer'
    ORDER BY country ASC, total DESC
""")
```

6. 查询 2024 年 3 月份的订单(order_date 在 '2024-03-01' 到 '2024-03-31' 之间),只显示 order_id, order_date, total,按日期升序。
提示:日期可以直接用字符串比较(因为是 'YYYY-MM-DD' 格式),BETWEEN 或者 >= <= 都行。

```python
duckdb.sql("""
    SELECT order_id, order_date, total
    FROM '../data/sales.csv'
    WHERE order_date BETWEEN '2024-03-01' AND '2024-03-31'
    ORDER BY order_date 
""")
```

建议

```python
# 看一下日期的最小值最大值
duckdb.sql("""
     SELECT MIN(order_date) AS first_day, MAX(order_date) AS last_day, COUNT(*) AS total
     FROM '../data/sales.csv'
""")
```

7. 查询所有产品名以 'M' 开头的订单(Monitor、Mouse 等),显示 product, quantity, total,前 10 行。

```python
duckdb.sql("""
    SELECT product, quantity, total
    FROM '../data/sales.csv'
    WHERE product LIKE 'M%'
    LIMIT 10 
""")
```

建议

LIKE 'M%' 用得标准。补充一个细节:LIKE 默认大小写敏感(在不同数据库可能不同,DuckDB 是敏感的)。如果想忽略大小写:

```python
# WHERE LOWER(product) LIKE 'm%'
```

或者

```python
# WHERE product ILIKE 'm%'   -- I 是 case-Insensitive
```

8. 计算列:查询每个订单的 order_id, product, quantity, price, 以及单件税后价格(税率 13%,公式:price * 1.13),起别名为 price_with_tax。只显示前 5 行。
💡 这道题练的是 Day 4 跟练 Part 2 —— SQL 里"列可以是表达式"。

```python
duckdb.sql("""
    SELECT order_id, product, quantity, price, price * 1.13 AS price_with_tax
    FROM '../data/sales.csv'
    LIMIT 5
""")
```

Hard

9. Top N 经典题:查询消费金额最高的 5 个订单,显示 order_id, customer_id, country, total。
💡 这道题在面试里被反复问。SQL 的写法是 ORDER BY ... DESC LIMIT N,记住这个模式。

```python
duckdb.sql("""
    SELECT order_id, customer_id, country, total 
    FROM '../data/sales.csv'
    ORDER BY total DESC
    LIMIT 5
""")
```

10. 组合查询:查询满足以下所有条件的订单:

country 是 UK 或 Germany 之一

total 大于等于 2000

category 不是 'Accessory'(用 != 或 <> 或 NOT IN)

按 total 降序排列

只显示前 10 行

显示列:order_id, country, category, product, total。

💡 这道题在测试你能不能把多个条件用 AND 串起来——实际工作中经常一个 WHERE 里挂 5-10 个条件。

```python
duckdb.sql("""
    SELECT order_id, country, category, product, total 
    FROM '../data/sales.csv'
    WHERE country IN ('UK', 'Germany') AND total >= 2000 AND category != 'Accessory'
    ORDER BY total DESC
    LIMIT 10
""")
```
