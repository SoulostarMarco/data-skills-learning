# Python Dict 与 Set（练习）

```python
# 1. dict 基础
# 创建dict
student = {'name': 'Alice', 'age': 20, 'major': 'CS'}
empty = {} # 空 dict
from_pairs = dict([('a', 1), ('b', 2)]) # 从键值对列表创建

print(from_pairs)

# 访问值
print(student['name']) # Alice
# print(student['height']) # KeyError

# 安全访问：用.get()，取不到返回None（或自定义默认值）
print(student.get('height')) # None
print(student.get('height', 170)) # 170（制定默认值）

# 修改/添加
student['age'] = 21
student['height'] = 165

# 删除
del student['major']
# 或用 pop，会返回被删除的值
popped = student.pop('height')

# 检查key是否存在
print('name' in student) # True
print('height' in student) # False
```

```python
# 2. dict的三种遍历方式
prices = {'apple': 5, 'banana': 3, 'cherry': 10}

# 遍历key
for fruit in prices: # 默认就是遍历key
    print(fruit)

# 遍历value
for price in prices.values():
    print(price)

# 遍历key+value
for fruit, price in prices.items():
    print(f'{fruit}: ${price}')
```

```python
# 3. dict的两个核心模式
# 模式1：计数
# 场景：统计list中每个元素出现次数
words = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
print(counts) # {'apple': 3, 'banana': 2, 'cherry': 1}

# 模式2：分组
# 场景：把学生按专业分组
students = [
    {'name': 'Alice', 'major': 'CS'},
    {'name': 'Bob', 'major': 'Math'},
    {'name': 'Charlie', 'major': 'CS'},
    {'name': 'David', 'major': 'Math'},
]
groups = {}
for s in students:
    major = s['major']
    if major not in groups:
        groups[major] = []
    groups[major].append(s['name'])
print(groups) # {'CS': ['Alice', 'Charlie'], 'Math': ['Bob', 'David']}
```

```python
# 4. 字典推导式（类似列表推导，但生成dict）
# 创建一个“数字 → 平方” 的映射
squares = {n: n**2 for n in range(1, 6)}
print(squares) # {1: 1, 2: 4, 3: 9, 4: 16, 5:25}

# 翻转key和value
prices = {'apple': 5, 'banana': 3}
reversed_prices = {v: k for k, v in prices.items()}
print(reversed_prices) # {5: 'apple', 3: 'banana'}

# 带条件
expensive = {k: v for k, v in prices.items() if v > 4}
print(expensive) # {'apple': 5}
```

```python
# 5. set基础
# 创建set
s = {1, 2, 3, 4}
empty_set = {} # 空set要这样建
from_list = set([1, 2, 2, 3, 3]) # {1, 2, 3} —— 自动去重

print(from_list)

# 添加/删除
s.add(5)
print(s)
s.discard(3) # 不存在不报错
print(s)
# s.remove(99) # 不存在会报错

# set的核心优势：O(1)查找
big_list = list(range(1000000))
big_set = set(big_list)

# 999999 in big_list → 要遍历100万次
# 999999 in big_set → 1次搞定（哈希查找）
```

```python
# 6. set运算
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(a & b) # 交集 {4, 5}
print(a | b) # 并集 {1, 2, 3, 4, 5, 6, 7, 8}
print(a - b) # 差集（在a中但不在b中）{1, 2, 3}
print(a ^ b) # 对称差集（只在其中一个中）{1, 2, 3, 6, 7, 8}

# 实际应用： 找出”新增用户“
yesterday_users = {'u1', 'u2', 'u3', 'u4'}
today_users = {'u2', 'u3', 'u5', 'u6'}

new_users = today_users - yesterday_users # {'u5', 'u6'}
churned = yesterday_users - today_users # {'u1', 'u4'}
retained = today_users & yesterday_users # {'u2', 'u3'}

print(f"新增：{new_users}，流失：{churned}，留存：{retained}")
```
