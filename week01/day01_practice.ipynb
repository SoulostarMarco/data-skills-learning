# Python List 与 Tuple (练习)

```python
# 1.创建与基础操作
nums = [3, 1, 4, 1, 5, 9, 2, 6]

print(len(nums)) # 长度
print(nums[0], nums[-1]) # 首尾部元素
print(nums[2:5]) # 切片 [start: end)， 不包含end
print(nums[::-1]) # 反转
print(nums[::2]) # 步长为2
```

```python
# 2. 修改list(list是可变的)
nums.append(7) # 末尾添加
print(nums)
nums.insert(0, 99) # 指定位置插入
print(nums)
nums.remove(1) # 删除第一个值为1的元素
print(nums)
popped = nums.pop() # 弹出末尾元素并返回
print(popped)
popped_first = nums.pop(0) # 弹出指定索引
print(popped_first)
```

```python
# 3.排序的两种方式
a = [3, 1, 4, 1, 5]
a.sort() # 原地排序， a本身被修改， 返回None
print(a)

b = [3, 1, 4, 1, 5]
b_sorted = sorted(b)
print(b, b_sorted)

# 降序
sorted([3, 1, 4], reverse=True)
```

```python
# 4.tuple：不可变的list
point = (3, 4)
# point[0] = 5 # 报错！tuple不能修改
x, y = point

# 为什么用tuple？
# - 可以作为dict的key（list不行）
# - 表示“一组固定的东西”，语义更清晰（比如坐标、RGB）
# - 函数返回多个值时， 本质是返回tuple

def min_max(lst):
    return min(lst), max(lst) # 返回 tuple

lo, hi = min_max([3, 1, 4, 1, 5])
print(lo, hi)
```

```python
# 5. list是引用类型
a = [1, 2, 3]
b = a # b 和 a指向同一个list
b.append(4)
print(a) # [1, 2, 3, 4] —— a 也变了

# 正确的复制方法
c = a.copy() # 或 c = a[:] 或 c = list(a)
c.append(99)
print(a) # 不受影响
```
