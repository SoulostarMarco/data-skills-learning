6. 给定一个订单列表,按用户分组(返回 dict: 用户 → 该用户所有订单金额的 list)。

```python
orders = [
       {'user': 'Alice', 'amount': 100},
       {'user': 'Bob',   'amount': 50},
       {'user': 'Alice', 'amount': 200},
       {'user': 'Charlie', 'amount': 75},
       {'user': 'Bob', 'amount': 150},
   ]
# 期望输出:{'Alice': [100, 200], 'Bob': [50, 150], 'Charlie': [75]}

groups = {}
for s in orders:
    user = s['user'] 
    if user not in groups:
        groups[user] = []
    groups[user].append(s['amount'])

print(groups)
```

参考答案

```python
from collections import defaultdict

groups = defaultdict(list)   # 关键:指定默认值的"工厂"是 list
for s in orders:
    groups[s['user']].append(s['amount'])

print(dict(groups))  # 转回普通 dict 打印更好看
```

defaultdict(list) 的意思是:"如果 key 不存在,自动给我创建一个空 list"——省掉了 if user not in groups: 这一步。
defaultdict 在数据处理里超级常用,记住这两种形式:

defaultdict(list) —— 默认空 list,用来分组

defaultdict(int) —— 默认 0,用来计数(可以替代 .get(k, 0)+1)

7. 接上题,进一步算出每个用户的总消费额(返回 dict: 用户 → 总金额),按总额降序排序,输出 list of tuple。

```python
# 期望输出:[('Alice', 300), ('Bob', 200), ('Charlie', 75)]
orders = [
       {'user': 'Alice', 'amount': 100},
       {'user': 'Bob',   'amount': 50},
       {'user': 'Alice', 'amount': 200},
       {'user': 'Charlie', 'amount': 75},
       {'user': 'Bob', 'amount': 150},
   ]

groups = {}
for s in orders:
    user = s['user'] 
    if user not in groups:
        groups[user] = s['amount']
    else:
        groups[user] += s['amount']
    
result = sorted(groups.items(), key=lambda x: x[1], reverse=True)
print(result)
```

8. 用字典推导式,把 ['apple', 'banana', 'cherry'] 转成 {'apple': 5, 'banana': 6, 'cherry': 6}(value 是单词长度)。


```python
fruits = ['apple', 'banana', 'cherry']
fruits_dict = {}

for fruit in fruits:
    fruits_dict[fruit] = fruits_dict.get(fruit, len(fruit))

print(fruits_dict)
```

参考答案

```python
fruits = ['apple', 'banana', 'cherry']
fruits_dict = {fruit: len(fruit) for fruit in fruits}
print(fruits_dict)
```

题目要求用字典推导式,你用了普通 for 循环 + .get()。而且 fruits_dict.get(fruit, len(fruit)) 这里的 .get() 用得不对——因为 fruits 里没有重复,所以 key 永远不存在,永远走 default 分支,逻辑上等价于直接赋值。

记住字典推导的语法结构:{key表达式: value表达式 for 变量 in 序列}

9. 给定一段英文文本,找出出现次数最多的 3 个单词(忽略大小写)。

```python
text = "the quick brown fox jumps over the lazy dog the fox is quick and the dog is lazy"
# 期望输出类似:[('the', 4), ('quick', 2), ('fox', 2)]
# 提示:text.split() 可以按空格切分;排序用 sorted(..., key=...)

text_lst = text.lower().split()
counts = {}

for word in text_lst:
    counts[word] = counts.get(word, 0) + 1

counts = list(counts.items())
print(sorted(counts, key=lambda x: x[1], reverse=True)[:3])
```

10. 两个 list 求交集,保留重复次数。
给定 a = [1, 2, 2, 1, 3] 和 b = [2, 2, 1, 4],返回[1, 2, 2] (顺序不要求)。
说明:1 在 a 中出现 2 次、在 b 中出现 1 次 → 取较小的 1 次;2 在 a 中 2 次、在 b 中 2 次 → 取 2 次。
提示:用两个 dict 分别计数,再对每个 key 取 min。

```python
a = [1, 2, 2, 1, 3]
b = [2, 2, 1, 4]
inter = set(a) & set(b)

counts_a = {}
counts_b = {}
final_lst = []

for num in a:
    counts_a[num] = counts_a.get(num, 0) + 1

for num in b:
    counts_b[num] = counts_b.get(num, 0) + 1

for i in inter:
    if counts_a[i] >= counts_b[i]:
        for _ in range(counts_b[i]):
            final_lst.append(i)
    else:
        for _ in range(counts_a[i]):
            final_lst.append(i)

print(final_lst)
```

参考答案

```python
a = [1, 2, 2, 1, 3]
b = [2, 2, 1, 4]

counts_a, counts_b = {}, {}
for x in a: counts_a[x] = counts_a.get(x, 0) + 1
for x in b: counts_b[x] = counts_b.get(x, 0) + 1

result = []
for k, va in counts_a.items():
    vb = counts_b.get(k, 0)
    result.extend([k] * min(va, vb))

print(result)
```

```python
result = [k for k, va in counts_a.items() 
          for _ in range(min(va, counts_b.get(k, 0)))]
print(result)
```

你的代码逻辑完全正确,但有冗余:

冗余 1:你已经用 inter = set(a) & set(b) 拿到了交集 key,然后又用 if counts_a[i] >= counts_b[i]: ... else: ... 手动取小。直接用内置的 min() 即可

冗余 2:for _ in range(n): lst.append(i) 可以直接写成 lst.extend([i] * n),或者列表推导。

冗余 3:其实根本不需要先求 inter——直接遍历 counts_a,遇到 counts_b 里没有的 key 就 .get(k, 0) 返回 0,自动跳过(因为 min(x, 0) = 0,乘 0 就是空 list)。
