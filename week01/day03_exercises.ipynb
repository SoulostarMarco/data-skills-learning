5. 给定一个二维矩阵 matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]:

5a. 用列表推导式扁平化成 [1, 2, 3, 4, 5, 6, 7, 8, 9]

5b. 用列表推导式提取对角线元素 [1, 5, 9](提示:用 enumerate)

5c. 用列表推导式把矩阵转置成 [[1, 4, 7], [2, 5, 8], [3, 6, 9]](进阶,有难度可跳过)

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [x for sub in matrix for x in sub]
print(flat)
```

```python
diagonal = [x for i, row in enumerate(matrix) for j, x in enumerate(row) if i == j]
print(diagonal)
```

参考答案

```python
[row[i] for i, row in enumerate(matrix)]
```

逻辑上没问题,但其实可以更直接——既然只要 i == j 的位置,直接用一层循环 + 索引访问就行

```python
[matrix[i][i] for i in range(len(matrix))]
```

或者更简洁(假设矩阵是方阵)

```python
transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(transposed)
```

```python
list(map(list, zip(*matrix)))
```

zip(*matrix) 用解包 + zip 直接实现转置。等学了 zip 你会回头看这个

6. 给定学生数据:

6a. 按分数降序排序,返回排序后的 list

6c. 用列表推导式,提取所有 A 班学生的名字 → ['Alice', 'Charlie', 'Eve']

```python
result = sorted(students, key=lambda x: x['name'], reverse=True)   # 先排次要的
result = sorted(result, key=lambda x: x['class'])                   # 再排主要的
```

7. 给定一个温度数据(摄氏度):temps_c = [0, 15, 23, 30, -5, 40]

7a. 用列表推导式,把所有温度转成华氏度(公式:F = C * 9/5 + 32)

7b. 在 7a 基础上,只保留华氏度 > 80 的温度,且保留摄氏度和华氏度的 tuple:

期望输出形如 [(30, 86.0), (40, 104.0)]

```python
temps_c = [0, 15, 23, 30, -5, 40]
print([C * 9/5 + 32 for C in temps_c])
```

```python
print([(C, C * 9/5 + 32) for C in temps_c if C * 9/5 + 32 > 80])
```

```python
print([(C, F) for C in temps_c for F in [C * 9/5 + 32] if F > 80])
```

小优化(可读性):7b 里你计算了两次 C * 9/5 + 32(一次在过滤,一次在 tuple 里),逻辑上重复。可以用一个"前置变换"避免

8. 给定一段文本:

8a. 用列表推导式,返回所有长度 >= 4 的单词

8b. 返回每个单词的 (单词, 长度) tuple,按长度降序排序,长度相同时按字母升序

期望输出:['A005', 'A003', 'A001']
提示:列表推导式 + sorted 嵌套,先想清楚顺序。

```python
orders = [
       {'order_id': 'A001', 'user': 'Alice', 'amount': 120, 'status': 'paid'},
       {'order_id': 'A002', 'user': 'Bob',   'amount': 80,  'status': 'pending'},
       {'order_id': 'A003', 'user': 'Alice', 'amount': 200, 'status': 'paid'},
       {'order_id': 'A004', 'user': 'Charlie', 'amount': 50, 'status': 'paid'},
       {'order_id': 'A005', 'user': 'Bob',   'amount': 300, 'status': 'paid'},
   ]

orders_filter = [order for order in orders if order['status'] == 'paid' and order['amount'] > 100]
print([order['order_id'] for order in sorted(orders_filter, key= lambda x: x['amount'], reverse=True)])
```

10. 二维数据透视(挑战题):

给定一个销售记录 list, 要求生成一个 dict,**key 是月份,value 是该月份每个产品的总销售额(dict 嵌套 dict)**:

提示:这其实是 Pandas `pivot_table` 的雏形。允许用 for 循环(因为嵌套 dict 用推导式会很难读),但要尽量简洁。

进阶:能不能用 `defaultdict` 让代码更优雅?(选做)

```python
sales = [
        ('2024-01', 'A', 100),
        ('2024-01', 'B', 150),
        ('2024-02', 'A', 120),
        ('2024-02', 'B', 80),
        ('2024-02', 'A', 30),   # 同一月份同一产品可能有多条
        ('2024-03', 'B', 200),
    ]

sales_dict = {}

for month, product, amount in sales:
    if month not in sales_dict:
        sales_dict[month] = {}
    if product not in sales_dict[month]:
        sales_dict[month][product] = 0
    sales_dict[month][product] += amount

print(sales_dict)
```

```python
from collections import defaultdict

# 嵌套 defaultdict:外层 dict 的默认值是 "内层 dict 默认值为 0 的 defaultdict"
sales_dict = defaultdict(lambda: defaultdict(int))

for month, product, amount in sales:
    sales_dict[month][product] += amount

# 转回普通 dict 打印
print({k: dict(v) for k, v in sales_dict.items()})
```

defaultdict(int) 的默认值是 0,所以可以直接 += amount,不需要先判断 key 存不存在。defaultdict(lambda: defaultdict(int)) 是嵌套版本,在数据岗里你会越来越常见。
