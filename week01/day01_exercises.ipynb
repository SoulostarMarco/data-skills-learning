1. 给定 nums = [5, 2, 8, 1, 9, 3],不使用 sort()/sorted(),找出最大值和最小值,返回 tuple (max, min)。

```python
nums = [5, 2, 8, 1, 9, 3]
print((max(nums), min(nums)))
```

参考答案

```python
# 使用循环的方法
nums = [5, 2, 8, 1, 9, 3]
mx, mn = nums[0], nums[0]
for num in nums:
    if num > mx:
        mx = num
    if num < mn:
        mn = num
print((mx, mn))
```

2. 给定一个 list ['apple', 'banana', 'cherry', 'date'],返回所有长度 > 5 的字符串组成的新 list。

```python
word_list = ['apple', 'banana', 'cherry', 'date']
new_list = []

for word in word_list:
    if len(word) > 5:
        new_list.append(word)

print(new_list)
```

参考答案

Python 里这种"过滤"操作几乎都会用列表推导式写

```python
word_list = ['apple', 'banana', 'cherry', 'date']
new_list = [word for word in word_list if len(word) > 5]
print(new_list)
```

4. 给定两个 list a = [1, 2, 3] 和 b = [4, 5, 6],合并成 [1, 2, 3, 4, 5, 6](至少想出 2 种方法)。

```python
a = [1, 2, 3]
b = [4, 5, 6]
# 方法1
union_ab_1 = a + b
print(union_ab_1)

# 方法2
union_ab_2 = a.copy()
for num in b:
    union_ab_2.append(num)
print(union_ab_2)
```

参考答案

第 4 题方法 2 有个小升级:用 extend 替代 for + append,语义更清晰

```python
# 方法2升级
union_ab_2 = a.copy()
union_ab_2.extend(b)
print(union_ab_2)
```

附赠第三种写法(unpacking,Python 3.5+)

```python
# 方法3
union_ab_3 = [*a, *b]
print(union_ab_3)
```

5. 给定 nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3],去重后返回新 list,顺序保持原样(即不能用 set 直接去重,因为会打乱顺序)。

```python
nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
unique_nums = []

for num in nums:
    if num not in unique_nums:
        unique_nums.append(num)

print(unique_nums)
```

参考答案

优化版用 set 做查重(set 的 in 是 O(1))

```python
nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
seen = set()
unique_nums = []
for num in nums:
    if num not in seen:
        seen.add(num)
        unique_nums.append(num)

print(unique_nums)
```

6. 给定一个学生成绩 list: students = [('Alice', 85), ('Bob', 72), ('Charlie', 90), ('David', 68)] 按成绩降序排序,返回排序后的 list。(提示:sorted 的 key 参数)

语法：sorted(iterable, key=None, reverse=False)

```python
students = [('Alice', 85), ('Bob', 72), ('Charlie', 90), ('David', 68)]
print(sorted(students, key=lambda x: x[1], reverse=True))
```

参考答案

对于 tuple,也可以用 operator.itemgetter(1),大数据下比 lambda 略快

```python
from operator import itemgetter
sorted(students, key=itemgetter(1), reverse=True)
```

7. 给定一个嵌套 list [[1, 2], [3, 4], [5, 6]],扁平化成 [1, 2, 3, 4, 5, 6]。

```python
nums = [[1, 2], [3, 4], [5, 6]]
flatten_nums = []

for i in range(len(nums)):
    for j in range(len(nums[i])):
        flatten_nums.append(nums[i][j])

print(flatten_nums)
```

参考答案

一行列表推导(注意 for 的顺序,从外到内)

```python
flatten = [item for sublist in nums for item in sublist]

print(flatten)
```

8. 实现一个函数 chunks(lst, n),把 list 切成长度为 n 的小块。

例如：chunks([1, 2, 3, 4, 5, 6, 7], 3) 返回 [[1, 2, 3], [4, 5, 6], [7]]

```python
def chunks(lst, n):
    chunk_lst = []
    num_lst = (len(lst)//n)
    remainder = len(lst)%n
    
    if remainder == 0:
        for i in range(num_lst):
            chunk_lst.append(lst[(i)*n:(i+1)*n])
    else:
        for i in range(num_lst):
            chunk_lst.append(lst[(i)*n:(i+1)*n])
        chunk_lst.append(lst[(num_lst)*n:])
    return chunk_lst

lst = [1, 2, 3, 4, 5, 6, 7]
chunks(lst, 3)
```

参考答案

分了余数和非余数两种情况——其实没必要,Python 切片越界不会报错,会自动截断；利用这点,可以一行解决

```python
def chunks(lst, n):
    return [lst[i:i+n] for i in range(0, len(lst), n)]

lst = [1, 2, 3, 4, 5, 6, 7]
chunks(lst, 3)
```

9. 给定一个 list [1, 2, 2, 3, 3, 3, 4],返回每个元素出现的次数,格式:[(1, 1), (2, 2), (3, 3), (4, 1)]。不要用 Counter,用基础 list 操作实现。

```python
nums = [1, 2, 2, 3, 3, 3, 4]
unique_nums = []
counters = []
combined_lst = []


for num in nums:
    if num not in unique_nums:
        unique_nums.append(num)

for unique_num in unique_nums:
    counter = 0
    for i in range(len(nums)):
        if unique_num == nums[i]:
            counter += 1
    counters.append(counter)

for j in range(len(unique_nums)):
    combined_lst.append((unique_nums[j], counters[j]))

print(combined_lst)
```

参考答案

思路是:先找 unique → 再数次数 → 再合并。逻辑清晰但效率低(O(n²) 而且遍历三次)。用 dict 一遍搞定

```python
nums = [1, 2, 2, 3, 3, 3, 4]
counts = {}
for num in nums:
    counts[num] = counts.get(num, 0) + 1
result = list(counts.items())
print(result)  # [(1, 1), (2, 2), (3, 3), (4, 1)]
```

两个关键点:

1. dict.get(key, default) —— 取不到 key 时返回 default,避免 KeyError,这是 Python 极常用技巧
2. Python 3.7+ 的 dict 保持插入顺序,所以输出顺序和原 list 首次出现顺序一致

10. 给定股票价格 list prices = [7, 1, 5, 3, 6, 4],求一次买卖能获得的最大利润(只能买一次卖一次,且必须先买后卖)。 例:第 2 天(价格 1)买,第 5 天(价格 6)卖,利润 = 5。

```python
prices = [7, 1, 5, 3, 6, 4]
profits = []

for i in range(len(prices)):
    for j in range(i+1, len(prices)):
        profit = prices[j] - prices[i]
        profits.append(profit)

print(max(profits))
```

参考答案

解法是 O(n²) 双重循环,能 AC,但面试官会追问"能不能更快"。这道题(LeetCode 121)的标准最优解是 O(n) 一次遍历:

```python
def max_profit(prices):
    if not prices:
        return 0
    min_price = prices[0]
    best = 0
    for p in prices[1:]:
        best = max(best, p - min_price)   # 假设今天卖,看看赚多少
        min_price = min(min_price, p)     # 更新到今天为止的最低买入价
    return best

prices = [7, 1, 5, 3, 6, 4]
print(max_profit(prices))
```
